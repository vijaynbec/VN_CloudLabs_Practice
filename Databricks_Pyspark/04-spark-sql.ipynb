```python
# dataframe method 1
fire_df = (spark.read
      .format("csv")
      .option("header", "true")
      .option("inferSchema", "true")
      .load("dbfs:/Volumes/workspace/sample/vol/Fire_Department_Calls_Service.csv"))

# display(fire_df)


#-----------Data frame to table

# Check total count from DF table
fire_df.createOrReplaceTempView("fire_table")
count = spark.sql("select count(*) from fire_table")
display(count)

# Check sample data from DF table
fire_df.createOrReplaceTempView("fire_table")
sample = spark.sql("select * from fire_table limit 10")
display(sample)

# Q1 How many distinct calls made to fire department - Call Type Group

calltypecount = spark.sql("select count(distinct(`Call Type Group`)) from fire_table where `Call Type Group` is not null")
display(calltypecount)

# Q2 Whate were distict call types

calltype = spark.sql("select distinct `Call Type Group` from fire_table where `Call Type Group` is not null")
display(calltype)

# Q3 most common call type count

calltypecount = spark.sql("select `Call Type Group`, count(*) as count from fire_table where `Call Type Group` is not null group by `Call Type Group` order by count desc")
display(calltypecount)

# Q4 most call type by zipcode
calltypecountbyzip = spark.sql("select  `Call Type Group`,`Zipcode of Incident`, count(*) as count from fire_table where `Call Type Group` is not null group by `Zipcode of Incident`, `Call Type Group` order by count desc")
display(calltypecountbyzip)

# Q5 neighborhood by zipcode 94012 and 94103
neighborhoodbyzipcode = spark.sql("select `Zipcode of Incident`, `Neighborhooods - Analysis Boundaries` from fire_table where `Zipcode of Incident` = '94012' or `Zipcode of Incident` = '94103'")
display(neighborhoodbyzipcode)

# Q6 number of alarams
alaramcount = spark.sql("select `Number of Alarms`, count(*) as count from fire_table where `Number of Alarms` is not null group by `Number of Alarms` order by count desc")
display(alaramcount)
```
