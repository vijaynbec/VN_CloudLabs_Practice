Each customer, produce the following the summary per month
- Total Orders
- Total items brought
- Total amount spent

```python
%sql
select customer_id,
       count(distinct item_id) as total_items,
       sum(quantity) as total_quanitity,
       round(sum(items_price*quantity),2) as total_price_customer,
       date_format(order_date, 'yyyy-MM') as order_month
from project_etl.silver.v_orders
group by customer_id, order_month;




```

```python
%sql
create or replace table project_etl.silver.orders_summary_silver
as
select customer_id,
       count(distinct item_id) as total_items,
       sum(quantity) as total_quanitity,
       round(sum(items_price*quantity),2) as total_price_customer,
       date_format(order_date, 'yyyy-MM') as order_month
from project_etl.silver.v_orders
group by customer_id, order_month;
```
