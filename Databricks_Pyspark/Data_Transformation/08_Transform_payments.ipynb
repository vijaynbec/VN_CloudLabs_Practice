```python
%sql
select payment_id,
payment_date as payment_timestamp,
date_format(payment_timestamp,"yyyy-MM-dd") as payment_date, -- comes as string
CAST(date_format(payment_timestamp,"yyyy-MM-dd") as date) as payment_date_updated, -- comes as date format
date_format(payment_timestamp,"HH:mm:ss") as payment_time,  -- comes as string
order_id,
customer_id,
amount,
payment_method,
status,
case status
  when "failed" then 1
  when "completed" then 0
  when "pending" then null
end as payment_status,
card_type,
transaction_reference from project_etl.bronze.v_payments;
```
