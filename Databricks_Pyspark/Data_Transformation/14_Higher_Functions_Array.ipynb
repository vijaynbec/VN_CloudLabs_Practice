Common used Array functions
- Transform
- Filter
- Exists
- Aggregator

    <function_name> (arrya_column, lambda_expression)

     lambda_expression: element => expression

```python
%sql
create or replace temporary view order_item as
select * from
values (1, array('phone','laptop', 'mouse')), 
(2,array('tv','car','bus')),
(3,array('office', 'chair', 'table','busstop')) as ordres(order_id,items);
```

```python
%sql
select * from order_item;
```

```python
%sql
select order_id, transform(items, x -> upper(x)) as upper_items from order_item;
-------x is a temporary variable to hold the field and lambda logic
```

```python
%sql
select order_id, filter(items, x -> x like 'bus%') as bus_items from order_item;
```

```python
%sql
select order_id, exists(items, x -> x = 'bus') as bus_items from order_item;
```

```python
%sql
create or replace temporary view order_item as
select * from
values 
(1, array(
  named_struct('name','laptop', 'price',6999),
  named_struct('name','mouse', 'price',69),
  named_struct('name','monitor', 'price',999))),
(2, array(
  named_struct('name','laptop2', 'price',6999),
  named_struct('name','mouse2', 'price',69),
  named_struct('name','monitor2', 'price',999))),
(3, array(
  named_struct('name','laptop3', 'price',6999),
  named_struct('name','mouse3', 'price',69),
  named_struct('name','monitor3', 'price',999)))
 as ordres(order_id,items);
```

```python
%sql
select * from order_item;
```

```python
%sql
select order_id, 
      transform (items, x -> named_struct (
                              'name', upper(x.name),
                              'price', round(x.price * 1.10, 2) 
                               )) items_with_tax from order_item;
```

```python
%sql
select order_id,
aggregate(items, 0, (cost, x) -> cost + x.price) as total_price from order_item;
```
