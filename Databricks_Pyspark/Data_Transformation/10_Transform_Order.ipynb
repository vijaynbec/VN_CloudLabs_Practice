```python
%sql
SELECT
  customer_id,
  item.item_id,
  item.price AS items_price,
  item.quantity,
  item.product_detail.brand AS product_brand,
  item.product_detail.product_name AS product_name,
  item.product_detail.color AS product_color,
  order_date,
  total_amount as total_oder,
  item.quantity * item.price AS item_cost
FROM project_etl.bronze.v_orders
LATERAL VIEW explode(items) AS item
-- LATERAL VIEW explode(items) AS item: 
-- This expands each element of the 'items' array into a separate row, 
-- allowing access to individual item fields for each order.
```
