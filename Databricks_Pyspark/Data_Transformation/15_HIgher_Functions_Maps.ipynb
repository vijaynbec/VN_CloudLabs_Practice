Common used map functions
- Transform_values
- Transform_keys
- map_filter

    <function_name> (map_column, lambda_expression)

     lambda_expression: (key, value) => expression

```python
%sql
create or replace temporary view order_item_map as
select * from
values (1, map('phone',1000,'laptop', 9595, 'mouse',85)), 
(2,map('tv',2500,'car',8555,'bus',9900)),
(3,map('office',150, 'chair',2500, 'table',500,'busstop',50)) as ordres(order_id,items);
```

```python
%sql
select * from order_item_map;
```

```python
%sql
select order_id,
transform_keys(items,(item,price) -> upper(item)) as upper_items from order_item_map;
```

```python
%sql
select order_id,
transform_values(items, (item,price) -> ROUND(price*1.10,2)) as new_price from order_item_map;
```

```python
%sql
select order_id,
map_filter(items, (item,price) -> price > 5000 ) as filter_price from order_item_map;
```
