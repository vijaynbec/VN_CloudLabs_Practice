```python
%sql
select * from
(select address_id,
 address_type,
 street_address, 
 city, 
 state, 
 postal_code,
 country 
 from 
 project_etl.bronze.v_address)
  pivot (max(street_address) as `street address`,
         max(city) as city,
         max(state) as state,
         max(postal_code) as postal_code,
         max(country) as country
         for address_type in ("billing", "shipping"));
```

```python
%sql
create or replace table project_etl.silver.address_silver as
select * from (
  select address_id,
         address_type,
         street_address, 
         city, 
         state, 
         postal_code,
         country 
    from project_etl.bronze.v_address
)
pivot (
  max(street_address) as street_address,
  max(city) as city,
  max(state) as state,
  max(postal_code) as postal_code,
  max(country) as country
  for address_type in ("billing", "shipping")
);
```
