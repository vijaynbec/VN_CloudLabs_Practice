```python
%sql
DESCRIBE FUNCTION EXTENDED project_etl.default.get_capital_custname;
```

```python
%sql
create or replace function project_etl.default.get_payment_status(status string) 
returns string 
return case status
  when "failed" then 1
  when "completed" then 0
  when "pending" then null
  end;
```

```python
%sql DESCRIBE FUNCTION EXTENDED project_etl.default.get_payment_status;  
```
