```python
%sql
select c.customer_id,
       c.customer_name,
       c.email,
       c.date_of_birth,
       c.member_since,
       a.billing_street_address,
       a.billing_city,
       a.billing_state,
       a.billing_postal_code,
       a.billing_country,
       a.shipping_street_address,
       a.shipping_city,
       a.shipping_state,
       a.shipping_postal_code,
       a.shipping_country
from project_etl.silver.customers_silver c
inner join project_etl.silver.address_silver a on c.customer_id = a.address_id;
```

```python
%sql
create or replace table project_etl.gold.customer_address_gold
as
select c.customer_id,
       c.customer_name,
       c.email,
       c.date_of_birth,
       c.member_since,
       a.billing_street_address,
       a.billing_city,
       a.billing_state,
       a.billing_postal_code,
       a.billing_country,
       a.shipping_street_address,
       a.shipping_city,
       a.shipping_state,
       a.shipping_postal_code,
       a.shipping_country
from project_etl.silver.customers_silver c
inner join project_etl.silver.address_silver a on c.customer_id = a.address_id;
```
