```python
# How many files are in the directory? and read one sample one and read selective columns and apply the transformation
display(dbutils.fs.ls('/databricks-datasets'))
display(dbutils.fs.ls('/databricks-datasets/airlines/part-00000'))

airlinesDF = spark.read\
    .format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("samplingRatio", "0.0002")    \
    .load("/databricks-datasets/airlines/part-00000")
## display thdataframe
#display(airlinesDF)
# Option to see selective columns by name

airlinesDF.select("Origin","Dest","Distance").show(10)

# option to see the columns by column name, dataframe and more
from pyspark.sql.functions import *
airlinesDF.select("Origin",col("Dest"),airlinesDF.Distance,"Year").show(10)


airlinesDF.select(
    "Origin",
    col("Dest"),
    expr("to_date(concat(Year, lpad(Month,2,'0'), lpad(DayofMonth,2,'0')), 'yyyyMMdd') as flightdate")
  ).show(10)



```
