```python
import dlt
from pyspark.sql.functions import current_timestamp, lit, col

@dlt.table(name="silver.customers_silver_dlt")
def customers_silver_dlt():
    return (
        dlt.read_stream("fivetran_source.customers_source_dlt")
         # ✅ Use dlt.read_stream to register dependency
            .withColumn("processing_time", current_timestamp())
            .withColumn("fivetran_source_file", col("_file"))
            .withColumn("source_table", lit("customers_silver_dlt"))
    )
```
