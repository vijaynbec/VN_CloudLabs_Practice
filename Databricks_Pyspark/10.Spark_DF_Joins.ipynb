```python
#--- Dataframe joins

from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *

#--- Join two DataFrames
#- Inner join
#- Outer join
#- Left join
#- Right join
#- Full join

# Order Data 
order_list = [("85","101", 350, 5000),
             ("86","102", 400, 6000),
             ("87","103", 450, 7000),
             ("88","104", 500, 8000),
             ("89","105", 550, 9000),
             ("90","106", 600, 10000),
             ("91","107", 650, 11000)]
order_df = spark.createDataFrame(order_list, ["order_id", "product_id", "quantity", "price"])
product_list = [("101", "Product1", "USA",100 ),
                ("102", "Product2", "Canada",200),
                ("103", "Product3", "Mexico",300),
                ("104", "Product4", "USA",400),
                ("105", "Product5", "Canada",500),
                ("120", "Product6", "Mexico",600),
                ("110", "Product7", "USA",700)]

product_df = spark.createDataFrame(product_list, ["product_id", "product_name", "country", "price"])

# show the dataframe
order_df.show()
product_df.show()

#--- Inner join
#- Inner join is the default join type
#- Inner join returns only those records that match in both DataFrames
#- Inner join is also called equi join
#- Inner join is also called natural join

join_expr = order_df.product_id == product_df.product_id


inner_df = order_df.join(product_df, join_expr, "inner")
inner_df.show()

#### selective columns from the inner join tables gives error for common 
# 
# INFO: column names - it happens when we explicilty select the columns..becouse spark engine doesn't know which column to pick as it registers the column name to to column id...and we explistly select the column name

#  inner_df.select("order_id", "product_id", "quantity", "price", "product_name", "country", "price").show()

# EROOR Reference `product_id` is ambiguous, could be: [`product_id`, `product_id`]. SQLSTATE: 42704"Reference `product_id` is ambiguous, could be: [`product_id`, `product_id`]. SQLSTATE: 42704"

# rename the common column name based on table/df
# Inner join with aliasing to resolve ambiguity
joined_df = order_df.alias("o").join(
    product_df.alias("p"),
    col("o.product_id") == col("p.product_id"),
    "inner"
)

# Select columns, renaming as needed
result_df = joined_df.select(
    col("o.order_id"),
    col("o.product_id"),
    # col("o.quantity"),
    col("o.price").alias("order_price"),
    col("p.product_name"),
    col("p.country"),
    col("p.price").alias("product_price")
)

display(result_df)

# Outer join (Full Outer join)

joined_outerdf = order_df.alias("o").join(
    product_df.alias("p"),
    col("o.product_id") == col("p.product_id"),
    "outer"
)

# Select columns, renaming as needed
outer_df = joined_outerdf.select("*")

display(outer_df)

# Outer join (Left Outer join)

joined_outerdf = order_df.alias("o").join(
    product_df.alias("p"),
    col("o.product_id") == col("p.product_id"),
    "left"
)

# Select columns, renaming as needed
outer_df = joined_outerdf.select("*")

display(outer_df)

# Outer join (Right Outer join)

joined_outerdf = order_df.alias("o").join(
    product_df.alias("p"),
    col("o.product_id") == col("p.product_id"),
    "Right"
)

# Select columns, renaming as needed
outer_df = joined_outerdf.select("*")

display(outer_df)
```
