```sql
%python

dbutils.fs.ls("s3://data-databricks-sample-102426687040/IN/")


df_customers = spark.read.option("header", "true").csv("s3://data-databricks-sample-102426687040/IN/customers-100.csv")
display(df_customers)
# List of columns to rename (example: "Internal ID" to "internal_id")
df_customers_clean = (
    df_customers
    .withColumnRenamed("Internal ID", "internal_id")
    .withColumnRenamed("Customer Id", "customer_id")
    .withColumnRenamed("First Name", "first_name")
    .withColumnRenamed("Last Name", "last_name")
    .withColumnRenamed("Company", "company")
    .withColumnRenamed("City", "city")
    .withColumnRenamed("Country", "country")
    .withColumnRenamed("Phone 1", "phone_1")
    .withColumnRenamed("Phone 2", "phone_2")
    .withColumnRenamed("Subscription Date", "subscription_date")
)
# Save the cleaned DataFrame as a table
df_customers_clean.write.mode("overwrite").saveAsTable("personal_vijay.s3_schema_in.customers_outs3")

df_customers_clean.write.option("header", "true").mode("overwrite").csv("s3://data-databricks-sample-102426687040/OUT/")


df_customers_froms3 = spark.read.option("header", "true").csv("s3://data-databricks-sample-102426687040/OUT/")

display(df_customers_froms3)
```
