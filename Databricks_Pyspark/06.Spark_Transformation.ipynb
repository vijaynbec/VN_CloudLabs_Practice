```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *
####### Definition of function
def to_date_df(df, fmt, fld):
  return df.withColumn(fld, to_date(col(fld),fmt))

########## Defining the schema and rows

my_schema = StructType([
  StructField("ID", IntegerType(), True),
  StructField("Event_Date", StringType(), True)])

# adding my rows to data frame 
my_rows = [Row(1, "01/01/2025"),Row(2, "01/02/2025"),Row(3, "01/03/2025")]

# my_rdd = spark.sparkContext.parallelize(my_rows)
my_df = spark.createDataFrame(my_rows, my_schema)

# print dataframe
my_df.show()

# create a new_df to convert the Event_Date column to date type
new_df = to_date_df(my_df, "MM/dd/yyyy", "Event_Date")
new_df.show()
```
