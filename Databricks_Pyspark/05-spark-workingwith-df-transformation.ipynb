```python
# dataframe method 1
raw_firedf = (spark.read
      .format("csv")
      .option("header", "true")
      .option("inferSchema", "true")
      .load("dbfs:/Volumes/workspace/sample/vol/Fire_Department_Calls_Service.csv"))


# Fixing the Dataframe column names
renamed_fire_df = raw_firedf.withColumnRenamed("Call Number","CallNumber") \
.withColumnRenamed("Unit ID", "UnitID") \
.withColumnRenamed("Incident Number", "IncidentNumber") \
.withColumnRenamed("Call Date", "CallDate") \
.withColumnRenamed("Watch Date", "WatchDate") \
.withColumnRenamed("Received DtTm", "ReceivedDtTm") \
.withColumnRenamed("Entry DtTm", "EntryDtTm") \
.withColumnRenamed("Dispatch DtTm", "DispatchDtTm") \
.withColumnRenamed("Response DtTm", "ResponseDtTm") \
.withColumnRenamed("On Scene DtTm", "OnSceneDtTm") \
.withColumnRenamed("Transport DtTm", "TransportDtTm") \
.withColumnRenamed("Hospital DtTm" , "HospitalDtTm") \
.withColumnRenamed("Call Final Disposition", "CallFinalDisposition") \
.withColumnRenamed("Available DtTm", "AvailableDtTm") \
.withColumnRenamed("Zipcode of Incident", "ZipcodeofIncident") \
.withColumnRenamed("Station Area", "StationArea") \
.withColumnRenamed("Box", "Box") \
.withColumnRenamed("Original Priority", "OriginalPriority") \
.withColumnRenamed("Final Priority", "FinalPriority") \
.withColumnRenamed("Neighborhooods - Analysis Boundaries", "NeighborhoodsAnalysisBoundaries") \
.withColumnRenamed("RowID", "RowID") \
.withColumnRenamed("Call Type Group", "CallTypeGroup") \
.withColumnRenamed("Number of Alarms", "NumberOfAlarms") \
.withColumnRenamed("Unit Type", "UnitType") \
.withColumnRenamed("Unit sequence in call dispatch", "UnitSequenceInCallDispatch") \
.withColumnRenamed("Fire Prevention District", "FirePreventionDistrict") \
.withColumnRenamed("Supervisor District", "SupervisorDistrict") \
.withColumnRenamed("ALS Unit", "ALSUnit")

display(renamed_fire_df)
# print the dataframe schema
renamed_fire_df.printSchema()

# All time stamp fields are strings fiels after display, lets modify them all id date is also string we need to convert also format changes if any.
#  |-- CallNumber: integer (nullable = true)
#  |-- DispatchDtTm: string (nullable = true)
#  |-- ResponseDtTm: string (nullable = true) and more to this we need 
# to import spark function at first place

from pyspark.sql.functions import * 

converted_fire_df = renamed_fire_df.withColumn("CallDate", to_date(col("CallDate"), "MM/dd/yyyy")) \
.withColumn("ReceivedDtTm", to_timestamp(col("ReceivedDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("EntryDtTm", to_timestamp(col("EntryDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("DispatchDtTm", to_timestamp(col("DispatchDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("ResponseDtTm", to_timestamp(col("ResponseDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("OnSceneDtTm", to_timestamp(col("OnSceneDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("TransportDtTm", to_timestamp(col("TransportDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("HospitalDtTm", to_timestamp(col("HospitalDtTm"), "MM/dd/yyyy hh:mm:ss a")) \
.withColumn("AvailableDtTm", to_timestamp(col("AvailableDtTm"), "MM/dd/yyyy hh:mm:ss a"))

display(converted_fire_df)

# print the dataframe schema
converted_fire_df.printSchema()
## converted timestamp fields are now in timestamp format
 #|-- EntryDtTm: timestamp (nullable = true)
 #|-- DispatchDtTm: timestamp (nullable = true)
 #|-- ResponseDtTm: timestamp (nullable = true)
 #|-- OnSceneDtTm: timestamp (nullable = true)
 #|-- TransportDtTm: timestamp (nullable = true)
 #|-- HospitalDtTm: timestamp (nullable = true)

 # to run set of SQL's on same data frame..lets cash the DF

# converted_fire_df.cache()  # wont support in servless compute

# Q1 How many distinct calls made to fire department - Call Type Group
converted_fire_df.createOrReplaceTempView("fire_table")
# SQL method
calltypecount1 = spark.sql("select count(distinct(`CallTypeGroup`)) from fire_table where `CallTypeGroup` is not null")
display(calltypecount1)

# converted_fire_df.cache()  - this wont support in serverless compute
# Dataframe method
# calltypecount2 = converted_fire_df.select("CallTypeGroup") \
#                                         .distinct()\
#                                           .count() \
#        print(calltypecount2)

# Q2 Whate were distict call types

calltype = spark.sql("select distinct `CallTypeGroup` from fire_table where `CallTypeGroup` is not null")
display(calltype)


```
